Q2) analyse sales based on product categories and determine which category has the lowest sales and which has highest

Q4) analyze monthly profit from sales data and determine which month had the highest profit

Q6) Analyze sales and profit by customer segment

Q7) Analyze sales to profit ratio

```python
# Convert date column
data["Order Date"] = pd.to_datetime(  data["Order Date"] )
```

```python
data["Order Month"] = data["Order Date"].dt.month
data["Order Year"] = data["Order Date"].dt.year
data["Order Day of Week"] = data["Order Date"].dt.dayofweek
```

```python
#Q1) calculate the monthly sales of the store and identify which month has the highest sales and which month has lowest sales

sales_by_month = data.groupby("Order Month")["Sales"].sum().reset_index()
sales_by_month
```

```python
fig = px.line(sales_by_month,
        x = "Order Month",
        y = "Sales",
        title = "Monthly sales Analysis"
        )

fig.show()
```

```python
#Q2) analyse sales based on product categories and determine which category has the lowest sales and which has highest

sales_by_category = data.groupby("Category")["Sales"].sum().reset_index()
sales_by_category
```

```python
fig = px.pie(sales_by_category,
        values = "Sales",
        names = "Category",
        title = "sales_by_category"
        )

fig.show()
```

```python
# Q3) sales analysis based on sub-category
sales_by_subcategory = data.groupby("Sub-Category")["Sales"].sum().reset_index()
sales_by_subcategory
```

```python
fig = px.bar(sales_by_subcategory,
        x = "Sub-Category",
        y = "Sales",
        title = "Sales Analysis by Sub-category"
        )

fig.show()
```

```python
#Q4) analyze monthly profit from sales data and determine which month had the highest profit

profit_by_month = data.groupby("Order Month")["Profit"].sum().reset_index()
profit_by_month
```

```python
fig = px.line(profit_by_month,
        x = "Order Month",
        y = "Profit",
        title = "Monthly Profit Analysis"
        )

fig.show()
```

```python
#Q5) Analyze profit by category and sub-category
profit_by_category_subcategory = data.groupby(["Category", "Sub-Category"])["Profit"].sum().reset_index()
profit_by_category_subcategory
```

```python
fig = px.bar(
    profit_by_category_subcategory,
    x="Sub-Category",
    y="Profit",
    color="Category",
    title="Profit Analysis by Category and Sub-Category"
)
fig.show()
```

```python
#Q6) Analyze sales and profit by customer segment

sales_profit_by_segment = data.groupby("Segment").agg({ 'Sales' : 'sum' , 'Profit' : 'sum'  }).reset_index()
```

```python
fig = go.Figure()

fig.add_trace(  go.Bar(
    x = sales_profit_by_segment['Segment'],
    y = sales_profit_by_segment["Sales"],
    name = 'Sales'
   ) )

fig.add_trace(  go.Bar(
    x = sales_profit_by_segment['Segment'],
    y = sales_profit_by_segment["Profit"],
    name = 'Profit'
   ) )

fig.show()
```

```python
# Q7) Analyze sales to profit ratio

sales_profit_by_segment = data.groupby('Segment').agg({ 'Sales' : 'sum' , 'Profit' : 'sum' }).reset_index()
sales_profit_by_segment["Sales_to_profit_ratio"] = sales_profit_by_segment["Sales"] /  sales_profit_by_segment["Profit"]
print(sales_profit_by_segment[["Segment" , "Sales_to_profit_ratio"]])
```

 **Analysis Questions & Findings**
1. Monthly Sales Analysis  
→ November has the highest sales while February records the lowest, showing strong year-end seasonality.

2. Category-wise Sales Analysis  
→ Technology generates the highest revenue and Office Supplies the lowest.

3. Sub-Category Sales Analysis  
→ Sales are concentrated in top sub-categories like Phones and Chairs, while Fasteners and Labels contribute minimally.

4. Monthly Profit Analysis  
→ December is the most profitable month of the year.

5. Profit by Category and Sub-Category  
→ Technology leads total profit, but several sub-categories operate with weak or negative margins.

6. Segment-wise Sales and Profit Analysis  
→ The Consumer segment leads both sales and profit contribution.

7. Sales-to-Profit Ratio Analysis  
→ Profit efficiency varies by segment, with Home Office showing the strongest margins and Consumer the weakest.

 **Conclusion**
The analysis highlights strong seasonal sales growth, with year-end months
driving peak revenue and profit. Technology products and the Consumer segment
lead overall performance, but profit margins vary across categories and segments.
Improving pricing and discount strategies can enhance profitability
while focusing on high-performing products will support sustainable growth.
